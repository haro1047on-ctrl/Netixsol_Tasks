```python
# removing the unnamed index column
df = pd.read_csv('ships.csv')
df = df.drop(columns=[df.columns[0]])
# Cleaning up Whitespace & Special Characters
df = df.apply(lambda col: col.str.strip() if col.dtype == "object" else col)
df['Name'] = df['Name'].str.replace('LÃ¼deritz', 'Lüderitz', regex=False)

# Cleaning 'Launch Year' (Convert empty strings/spaces to actual NaN, then convert to numeric/nullable integer)
df['Launch Year'] = pd.to_numeric(df['Launch Year'].replace('', pd.NA), errors='coerce').astype('Int64')

# Handling Missing Values in Categorical Columns (Replacing empty class/country with 'Unknown')
df['Class'] = df['Class'].fillna('Unknown')
df['Country'] = df['Country'].fillna('Unknown')

# Removing all Duplicates
df = df.drop_duplicates()
# df.head(10)
df.tail(10)
```

4. Which country was building the widest variety of ship classes?

```python
df.groupby('Country')['Class'].value_counts()
```

```python
variety_by_country = df.groupby('Country')['Class'].nunique().sort_values(ascending=False)
print(variety_by_country)
```

**Answer:** The country with the highest number of widest ship varieties is USA.

5. What's the most common ship class for each of the top 3 countries?

```python
top_3_countries = variety_by_country.head(3).index.tolist()

print(top_3_countries)

# excluding 'No Classification'
for country in top_3_countries:
    country_df = df[df['Country'] == country]

    filtered_classes = country_df['Class'][country_df['Class'] != 'No Classification']

    if not filtered_classes.empty:
        most_common_class = filtered_classes.value_counts().idxmax()
        count_of_class = filtered_classes.value_counts().max()
        print(f"The most common ship class for {country} is '{most_common_class}' with {count_of_class} units.\n")
    else:
        print(f"No other classified ships found for {country} besides 'No Classification'.\n")
```

**Answer:** The most common ship class for each of the top 3 countries (United States, Japan, and Germany) is displayed above.

6. How much missing data is there in Launch Year, and does it differ by country?

```python
total_missing_launch_year = df['Launch Year'].isnull().sum()
print(total_missing_launch_year)
missing_launch_year_by_country = df.groupby('Country')['Launch Year'].apply(lambda x: x.isnull().sum()).sort_values(ascending=False)
display(missing_launch_year_by_country)
```

**Answer:** The total number of missing 'Launch Year' values is **717** and their distribution by country are displayed above. This shows how much missing data there is and **Germany** contributes the most to it as if we talk about the most missing values of launch year countrywise.

7. Which country had the earliest and latest ship launches on record?

```python
print(df['Launch Year'])
min_yr= df['Launch Year'].min()
print(min_yr)
max_yr= df['Launch Year'].max()
print(max_yr)

valid_launch_years = df['Launch Year'].dropna()
valid_launch_years = valid_launch_years[valid_launch_years != 0]

min_yr = valid_launch_years.min()
max_yr = valid_launch_years.max()
print('''
Min_yr & Max_yr after dropping null values:
''')
print(min_yr)
print(max_yr)

earliest_launch_country = df[df['Launch Year'] == min_yr]['Country'].iloc[0]
latest_launch_country = df[df['Launch Year'] == max_yr]['Country'].iloc[0]

print(earliest_launch_country,min_yr)
print(latest_launch_country,max_yr)
```

**Answer:** USA was the earliest & France is the Latest according to this dataset
