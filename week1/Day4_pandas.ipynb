**Checked missing values:**
all columns were checked so that it can be determined which to keep, which to drop and which to fix

```python

cat_cols = df.select_dtypes(include='object').columns
num_cols = df.select_dtypes(include=['float64', 'int64']).columns


df[cat_cols] = df[cat_cols].fillna('UNKNOWN')
df[num_cols] = df[num_cols].fillna(0)
```

Filling up the null values and fixing datatypes in next cell

Because:
- Most missing values here are **structurally missing, not errors**. For example, (Incendiary Devices) is blank
  when a mission didn't drop incendiary bombs, so it makes sense to fill **numeric columns with 0**.
- Text columns like (Aircraft Series) or (Target Type) are filled with ("UNKNOWN") instead of dropping,
  since dropping would throw away a huge portion of the dataset.
- Rows where the **Mission Date** itself is missing/unparsable are dropped, since date is essential for analysis
  and can't be reasonably guessed.

```python

df['Mission Date'] = pd.to_datetime(df['Mission Date'], errors='coerce')

numeric_like_cols = ['Takeoff Latitude', 'High Explosives Weight (Pounds)', 'Time Over Target']
for col in numeric_like_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

df = df.dropna(subset=['Mission Date'])

df.dtypes[['Mission Date', 'Takeoff Latitude', 'High Explosives Weight (Pounds)']]
```

**Fixed wrong Datatypes**

```python
df = df.drop_duplicates()
```

**Removed all duplicate rows from the dataset**

```python
df['Mission Year'] = df['Mission Date'].dt.year

df['Bombing Success Rate'] = df.apply(lambda row: row['Bombing Aircraft'] / row['Airborne Aircraft'] if row['Airborne Aircraft'] > 0 else 0,axis=1)

df[['Mission Date', 'Mission Year', 'Airborne Aircraft', 'Bombing Aircraft', 'Bombing Success Rate']].head()
```

**Created 2 Derived Columns**

- **Mission Year** is extracted from **Mission Date** using a vectorized **.dt** operation.
- **Bombing Success Rate** is the fraction of airborne aircraft that actually reached and bombed the target,
  calculated with **.apply()** since it needs conditional logic.

```python
summary = df.groupby(['Theater of Operations', 'Mission Year']).agg(Total_Missions=('Mission ID', 'count'),Total_Bomb_Weight_Tons=('Total Weight (Tons)', 'sum')).reset_index()
summary.head(10)
```

**Reshaped Data with groupby() + .agg()**

summarized the data by **Theater of Operations** and **Mission Year**, showing total missions flown
and total bomb weight dropped (in tons), turning 178k raw rows into a compact summary table.

```python
theater_lookup = pd.DataFrame({
    'Theater of Operations': ['ETO', 'PTO', 'MTO', 'CBI', 'EAST AFRICA', 'MADAGASCAR'],
    'Theater Full Name': [
        'European Theater of Operations',
        'Pacific Theater of Operations',
        'Mediterranean Theater of Operations',
        'China-Burma-India Theater',
        'East Africa Theater',
        'Madagascar Theater'
    ]
})

final_summary = summary.merge(theater_lookup, on='Theater of Operations', how='left')
final_summary
final_summary.head(11)
```
