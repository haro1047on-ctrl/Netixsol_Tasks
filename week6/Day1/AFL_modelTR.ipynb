```python
# pick one specific match that exists in both
sample_match = AFL_RBR_HA_df[['team','match_date']].drop_duplicates().iloc[10]
t, d = sample_match['team'], sample_match['match_date']

rbr_players = AFL_RBR_df[(AFL_RBR_df['team']==t) & (AFL_RBR_df['match_date']==d)]['player_id'].nunique()
rbr_ha_players = AFL_RBR_HA_df[(AFL_RBR_HA_df['team']==t) & (AFL_RBR_HA_df['match_date']==d)]['player_id'].nunique()

print(f'Match: {t} on {d}')
print('Players in original RBR file:', rbr_players)
print('Players in  RBR_HA file:', rbr_ha_players)
```

```python
# pick one specific match that exists in both
sample_match = AFL_RBR_HA_df[['team','match_date']].drop_duplicates().iloc[19]
t, d = sample_match['team'], sample_match['match_date']

rbr_players = AFL_RBR_df[(AFL_RBR_df['team']==t) & (AFL_RBR_df['match_date']==d)]['player_id'].nunique()
rbr_ha_players = AFL_RBR_HA_df[(AFL_RBR_HA_df['team']==t) & (AFL_RBR_HA_df['match_date']==d)]['player_id'].nunique()

print(f'Match: {t} on {d}')
print('Players in original RBR file:', rbr_players)
print('Players in  RBR_HA file:', rbr_ha_players)
```

```python
# for each unique match, compute % of players retained after merge
rbr_counts = AFL_RBR_df.groupby(['team','match_date'])['player_id'].nunique().rename('rbr_players')
rbr_ha_counts = AFL_RBR_HA_df.groupby(['team','match_date'])['player_id'].nunique().rename('rbr_ha_players')

compare = rbr_counts.to_frame().join(rbr_ha_counts, how='left').reset_index()
compare['rbr_ha_players'] = compare['rbr_ha_players'].fillna(0)
compare['retention_pct'] = compare['rbr_ha_players'] / compare['rbr_players']
compare['year'] = pd.to_datetime(compare['match_date']).dt.year
compare['decade'] = (compare['year']//10)*10

print(compare.groupby('decade')['retention_pct'].mean())
```
