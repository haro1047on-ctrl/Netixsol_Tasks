## Pre-Merge Data Quality Checks

Before combining the datasets, let's re-verify the cleanliness of `df1` and `df2`.

### Check for Missing Values in `df1` and `df2`

```python
print('Missing values in df1:')
display(df1.isnull().sum())

print('\nMissing values in df2 (top 10 by count):')
display(df2.isnull().sum().sort_values(ascending=False).head(10))
```

### Check for Duplicates in `df1` and `df2`

```python
print('Duplicates in df1:', df1.duplicated(subset=['player_id']).sum())
print('Duplicates in df2:', df2.duplicated().sum())
```

### Check Data Types, especially for the 'player_id' column

It's crucial that the `player_id` column has a consistent data type in both DataFrames for a successful merge. We previously saw a `DtypeWarning` for `df2['player_id']`, so let's convert it to numeric.

```python
# Convert 'player_id' in df2 to numeric, coercing errors to NaN
df2['player_id'] = pd.to_numeric(df2['player_id'], errors='coerce')

# Drop rows where 'player_id' became NaN after coercion (if any invalid entries existed)
df2.dropna(subset=['player_id'], inplace=True)

# Convert 'player_id' to integer type in both dataframes
df1['player_id'] = df1['player_id'].astype(int)
df2['player_id'] = df2['player_id'].astype(int)

print('Data types for df1:')
display(df1.dtypes)

print('\nData types for df2:')
display(df2.dtypes)
```

### Verify 'weight' column in `df1`

We previously handled zero values in the 'weight' column. Let's ensure it's free of invalid values and its statistics look reasonable.

```python
print('Description of weight column in df1 after cleaning:')
display(df1['weight'].describe())
```

### Verify 'games_played' column in `df2`

We removed negative `games_played` values. Let's ensure there are no more and its statistics are sound.

```python
print('Description of games_played column in df2 after cleaning:')
display(df2['games_played'].describe())
```

## Handling Missing Values in `df2`\n\nLet's calculate the percentage of missing values for each column in `df2` to decide which columns to keep.

```python
missing_percentages = (df2.isnull().sum() / len(df2)) * 100
missing_summary = pd.DataFrame({'Missing %': missing_percentages})
display(missing_summary.sort_values(by='Missing %', ascending=False).head(15))
```

## Combining the Datasets

I will now perform a `left join` to merge the seasonal statistics (`df2`) with the player background information (`df1`).

```python
combined_df = df2.merge(df1, on='player_id', how='left')

print(f'Shape of combined dataset: {combined_df.shape}')
display(combined_df.head())
```

```python
summary_table = combined_df.groupby(['team', 'decade']).agg({
    'kicks': 'mean',
    'disposal_efficiency': 'mean',
    'games_played': 'sum'
})
display(summary_table)
```

This summary shows how each team's average kicks and disposal efficiency have changed across decades, along with total games played by team per era
