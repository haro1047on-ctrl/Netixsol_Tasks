### Justification for Retaining NaN Values in Advanced Stat Columns

Several advanced statistical columns (e.g., `brownlow_votes`, `goal_assist`, `bounces`, `hit_outs`, etc.) exhibit a significant number of missing values. These missing values are not random but often stem from two primary reasons:

1.  **Historical Data Gaps:** Many of these statistics were not recorded or tracked consistently in earlier years of the AFL, particularly before the early 2000s. Imputing these values would involve making assumptions about player performance that cannot be reliably supported by historical data, potentially leading to inaccurate insights.
2.  **Player Roles/Game Situations:** Some stats are highly specific to certain player roles (e.g., `hit_outs` for ruckmen) or game situations. A player might legitimately have zero for a given stat in a game if their role did not involve it, or if the opportunity did not arise. Differentiating between a true zero and a missing data point is challenging, and forced imputation could misrepresent player contributions.

Therefore, retaining `NaN` values in these columns is the most appropriate approach. This preserves the integrity of the available data and avoids introducing artificial precision or bias through imputation. Any analysis involving these columns will need to account for their missingness, either by filtering, using imputation techniques that are specifically tailored to the context, or by acknowledging the limitations of the data for certain time periods.

```python
df2['goals_per_game'] = df2['goals'] / df2['games_played']
```

Why it matters: raw total goals rewards long careers as much as scoring skill

Already saw this exact issue with Tony Lockett's chart (career totals favor longevity). Dividing by games played gives a rate, so a player who scored efficiently in a short career isn't unfairly ranked below a long-career player with a lower actual scoring rate.

```python
df2['tackles_per_game'] = df2['tackles'] / df2['games_played']
```

Why it matters: tackles reflect defensive work rate/effort

A stat that's independent of scoring, so combined with goals it gives a more rounded picture of a player's overall contribution, not just their attacking output.

```python
df2['contested_poss_per_game'] = df2['contested_possessions'] / df2['games_played']
```

Why it matters: contested possessions (winning the ball under physical pressure, as opposed to uncontested/easy possessions) is generally considered a strong indicator of a player's competitiveness and ball-winning ability

A "hard at it" metric that raw disposal counts don't capture on their own.

```python
df2['score_involvement_per_game'] = (df2['goals'] + df2['goal_assists']) / df2['games_played']
```

Why it matters: goals alone only credits the player who kicked the ball through the posts

It misses players who set up scores (assists) without scoring themselves. Combining both gives a fuller picture of offensive contribution.

### Justification for `performance_index` Statistics

The `performance_index` is a composite metric designed to capture a player's overall on-field contribution by combining several key per-game statistics. The chosen statistics, derived from both raw game data and existing fantasy point systems, aim to provide a balanced view of offensive, defensive, and general utility:

1.  **`goals_per_game`**: This metric provides a rate-based measure of a player's primary scoring ability, normalizing for the number of games played. It moves beyond raw goal totals, which can favor players with long careers over those who score efficiently in fewer games.

2.  **`tackles_per_game`**: Reflecting a player's defensive work rate and effort, tackles are crucial for regaining possession and applying pressure. Including this statistic ensures that defensive contributions are recognized alongside offensive output, providing a more complete picture of a player's utility.

3.  **`contested_poss_per_game`**: Contested possessions are a strong indicator of a player's ability to win the ball under pressure, demonstrating competitiveness and ball-winning prowess. This 'hard at it' metric captures an aspect of player value that raw disposal counts alone do not.

4.  **`score_involvement_per_game`**: This combines goals and goal assists, offering a broader view of a player's offensive impact. It acknowledges players who set up scores (assists) as well as those who finish them, capturing their overall contribution to scoring plays.

5.  **`avg_fantasy_points`**: Fantasy scoring systems are specifically designed to weight multiple statistical categories (e.g., disposals, goals, tackles) into a single composite number. Therefore, `avg_fantasy_points` serves as a useful, pre-weighted summary metric that can act as a sanity check and provide a holistic view of a player's value based on a widely accepted scoring methodology.

```python
cols = ['goals_per_game','tackles_per_game','contested_poss_per_game','score_involvement_per_game','avg_fantasy_points']

for col in cols:
    df2[col + '_scaled'] = (df2[col] - df2[col].min()) / (df2[col].max() - df2[col].min())
```

```python
scaled_cols = [col + '_scaled' for col in cols]
df2['performance_index'] = df2[scaled_cols].mean(axis=1, skipna=False)
```

### Measuring Seasonal Performance Trends
defining **improvement** and **decline** using the Pearson correlation coefficient ($r$) between the `round_number` and `fantasy_points`.
- **Improvers ($r > 0$):** Players whose scores trend upwards over time.
- **Decliners ($r < 0$):** Players whose scores trend downwards over time.

*Note: We only consider player-seasons with at least 10 games for statistical significance.*

```python

trend_data = df1.dropna(subset=['round_number'])
game_counts = trend_data.groupby(['player_id', 'year'])['id'].transform('count')
trend_data = trend_data[game_counts >= 10]

trends = trend_data.groupby(['player_id', 'year']).apply(lambda x: x['round_number'].corr(x['fantasy_points']), include_groups=False).reset_index(name='trend_score')


trends = trends.merge(df2[['player_id', 'player_name']].drop_duplicates(), on='player_id')

print("Top 5 Improvers (Highest Positive Trend):")
display(trends.sort_values('trend_score', ascending=False).head(5))

print("\nTop 5 Decliners (Most Negative Trend):")
display(trends.sort_values('trend_score', ascending=True).head(5))
```

```python
improvers = trends.sort_values('trend_score', ascending=False).head(5)
decliners = trends.sort_values('trend_score', ascending=True).head(5)
plot_df = pd.concat([improvers, decliners])

plot_df['label'] = plot_df['player_name'] + ' (' + plot_df['year'].astype(str) + ')'

plt.figure(figsize=(10, 8))
colors = ['#2ecc71' if x > 0 else '#e74c3c' for x in plot_df['trend_score']]
sns.barplot(data=plot_df, x='trend_score', y='label', palette=colors)

plt.title('Seasonal Performance Trends: Top Improvers vs. Top Decliners', fontsize=14)
plt.xlabel('Trend Score (Correlation between Round and Fantasy Points)', fontsize=12)
plt.ylabel('Player (Season)', fontsize=12)
plt.axvline(0, color='black', lw=1)
plt.grid(axis='x', alpha=0.6)
plt.tight_layout()
plt.show()
```

### Explaining the Metrics
- **Peak Performance:** Measured by the highest `fantasy_points` score a player achieved in a single game within a season. This represents a player's absolute 'ceiling' or maximum potential impact.
- **Consistency (Volatility):** Measured by the standard deviation (`std`) of a player's scores. A lower value indicates a reliable, 'bankable' player, while a higher value indicates a 'boom-or-bust' performer.

```python
# 1. Distribution of Peak Scores
plt.figure(figsize=(10, 4))
sns.histplot(peak, kde=True, color='blue')
plt.title('Distribution of Player Peak Scores (Season Max)')
plt.xlabel('Maximum Fantasy Points in a Single Game')
plt.show()
```

#### Justification for Distribution Analysis
Analyzing the distribution of peak scores allows us to identify how rare elite-level performances are across the league. Similarly, the consistency distribution helps us understand the baseline 'reliability' of the average AFL player, providing a benchmark to identify truly exceptional outliers.

```python
# 2. Distribution of Consistency (Volatility)
# Extracting only the 'std' column from the consistency DataFrame for visualization
consistency_std = consistency['std']
plt.figure(figsize=(10, 4))
sns.histplot(consistency_std, kde=True, color='green')
plt.title('Distribution of Player Consistency (Standard Deviation)')
plt.xlabel('Standard Deviation of Fantasy Points')
plt.show()
```

```python
# 3. Relationship: Peak vs. Consistency
# Aligning peak (Series) and consistency (DataFrame) using a merge on player_id and year
peak_df = peak.reset_index(name='Peak')
metrics_df = peak_df.merge(consistency[['player_id', 'year', 'std']], on=['player_id', 'year'])

plt.figure(figsize=(10, 6))
sns.regplot(data=metrics_df, x='Peak', y='std',
            scatter_kws={'alpha':0.3, 's':10}, line_kws={'color':'red'})
plt.title('Relationship: Peak Performance vs. Scoring Volatility')
plt.xlabel('Season Peak (Max Score)')
plt.ylabel('Consistency (Std Dev)')
plt.show()
```

#### Defending the Peak vs. Consistency Relationship
This regression analysis is critical because it identifies the tradeoff between risk and reward. Traditionally, higher peak performers are expected to have higher volatility (less consistency). By plotting this relationship, can identify 'Unicorns': players who maintain a high peak while keeping their standard deviation low, representing the highest value assets for any team.
